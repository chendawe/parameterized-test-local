```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import TwoSlopeNorm

# 原始数据集
rows = ['GW150914', 'GW151012', 'GW151226', 'GW170104', 'GW170608',
        'GW170729', 'GW170809', 'GW170814', 'GW170818', 'GW170823']
cols = ['0', '2', '3', '4', '5l', '6', '6l']
data = [
    [0.7916, 0.6567, 1.154, -0.7571, 1.6963, -3.1731, -0.8042],
    [0.8702, 0.9543, 0.2644, 8.3517, 2., 1.0572, 0.7647],
    [0.9315, 1.6672, 0.1405, 4.6359, 3.8856, 0.1718, 2.9478],
    [0.7189, 0.8118, 0.8852, 0.4599, 1.5064, -7.031, -0.0205],
    [0.7991, 2.5007, -17.2831, 17.1035, 53.0508, -3.639, 6.9243],
    [0.5925, 0.9021, 0.7067, 1.2749, 0.2215, 0.224, 1.7035],
    [0.7948, 0.7376, 1.1812, -0.7041, 2.3158, 7.2182, -0.9733],
    [0.7944, 0.9641, 0.9009, 1.1593, 0.9252, 3.7164, 1.4542],
    [0.7853, 1.1693, 0.9343, 2.6573, -1.6771, 16.3357, -0.8567],
    [0.5917, 0.5392, 1.046, -0.2824, 1.4266, 13.3621, -0.4501]
]

new_rows = ['GW150914', 'GW170814', 'GW170608', 'GW170104', 'GW151226',
            'GW170809', 'GW170823', 'GW170818', 'GW170729', "GW151012"]

# 1. 数据重排序
row_to_data = dict(zip(rows, data))
new_data = [row_to_data[row] for row in new_rows]
df = pd.DataFrame(new_data, index=new_rows, columns=cols)

# 2. 核心变换：对数压缩量级 + 保留原始正负（适配median ratio=1为基准）
def transform_val(val):
    if val > 0:
        # 正数：以1为基准，val≥1→log10(val)；0<val<1→-log10(val)（保证<1为负，>1为正）
        return np.log10(val) if val >= 1 else -np.log10(val)
    elif val < 0:
        # 负数直接取对数后加负号（保持负向偏离）
        return -np.log10(abs(val) + 1e-6)
    else:
        return 0

# 应用变换
df_transformed = df.applymap(transform_val)

# 3. 关键调整：白色过渡点设为1（对应变换后的值为0，因为log10(1)=0）
# 计算相对于1的最大偏差，保证颜色对称
trans_center = 0  # 对应原始值1（log10(1)=0）
v_extreme = max(abs(df_transformed.values.min() - trans_center), 
                abs(df_transformed.values.max() - trans_center))
# vcenter=0 对应原始median ratio=1（白色），<0→蓝色（<1），>0→红色（>1）
norm = TwoSlopeNorm(vcenter=trans_center, vmin=-v_extreme, vmax=v_extreme)

# 4. 绘图（RdBu_r：<1→蓝，=1→白，>1→红）
plt.figure(figsize=(10, 10), dpi=100)
im = plt.imshow(df_transformed, cmap='RdBu_r', norm=norm, aspect='auto')

# 5. 标注原始median ratio值（保留4位小数）
for i in range(len(df.index)):
    for j in range(len(df.columns)):
        val_original = df.iloc[i, j]
        val_trans = df_transformed.iloc[i, j]
        if not np.isnan(val_original):
            # 深色背景用白色文字，浅色用黑色（保证可读性）
            text_color = 'white' if abs(val_trans) > 1 else 'black'
            plt.text(j, i, f'{val_original:.4f}', 
                     ha='center', va='center', 
                     color=text_color, fontsize=8)

# 6. 坐标轴与标签
plt.xticks(ticks=np.arange(len(cols)), labels=cols, fontsize=11)
plt.yticks(ticks=np.arange(len(new_rows)), labels=new_rows, fontsize=11)

# 7. 颜色条（明确说明白色对应median ratio=1）
cbar = plt.colorbar(im, shrink=0.8, pad=0.05)
cbar.set_label('Rescaled Median Ratio (1→White, <1→Blue, >1→Red)', fontsize=10)

# 8. 标题与布局
plt.title('Heat map of rescaled median ratio of $\\psi_{j(l)}$', fontsize=16, pad=20)
plt.xlabel('post-Newtonian order', fontsize=14)
plt.ylabel('Event ID', fontsize=14)

plt.tight_layout()
plt.savefig("/mnt/d/Documents/workspace/课题/论文/figs/ratios/heatmap_median_ratio_psi.pdf")
plt.show()
```
